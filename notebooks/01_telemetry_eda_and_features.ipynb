# FactoryPulse: Sensor Telemetry EDA & Feature Engineering

This notebook demonstrates physical telemetry analysis and temporal/spectral feature extraction across C-MAPSS, IMS Bearing, and AI4I industrial datasets.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load processed feature sets
df_cmapss = pd.read_parquet("data/processed/cmapss_features/train.parquet")
df_ims = pd.read_parquet("data/processed/ims_features/train.parquet")
df_ai4i = pd.read_parquet("data/processed/ai4i_features/train.parquet")

print(f"C-MAPSS shape: {df_cmapss.shape}")
print(f"IMS Bearing shape: {df_ims.shape}")
print(f"AI4I shape: {df_ai4i.shape}")
```

```python
# Plot C-MAPSS degradation trajectory
engine_1 = df_cmapss[df_cmapss["machine_id"] == "cmapss_1"]
plt.figure(figsize=(10, 4))
plt.plot(engine_1["operating_cycle"], engine_1["temperature"], label="T24 Temp", color="crimson")
plt.plot(
    engine_1["operating_cycle"],
    engine_1["vibration"] * 100,
    label="Vibration (scaled)",
    color="navy",
)
plt.title("Turbofan Engine #1 Degradation Trajectory")
plt.xlabel("Operating Cycle")
plt.ylabel("Sensor Readings")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
